```python
from scipy.stats import pearsonr

# Korelasi luas bangunan dengan harga
r_luas, p_luas = pearsonr(
    df["sqft_living"],
    df["price"]
)

print("Korelasi Luas Bangunan dengan Harga")
print("Koefisien korelasi:", r_luas)
print("p-value:", p_luas)


# Korelasi jumlah kamar dengan harga
r_kamar, p_kamar = pearsonr(
    df["bedrooms"],
    df["price"]
)

print("\nKorelasi Jumlah Kamar dengan Harga")
print("Koefisien korelasi:", r_kamar)
print("p-value:", p_kamar)
```

```python
korelasi = df[["price", "sqft_living", "bedrooms"]].corr()

print("Matriks Korelasi:")
print(korelasi)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    korelasi,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Matriks Korelasi Variabel Utama")
plt.show()
```

```python
from scipy.stats import linregress

# Regresi linear luas bangunan terhadap harga
reg_luas = linregress(df["sqft_living"], df["price"])

print("REGRESI LINEAR: LUAS BANGUNAN DAN HARGA")
print("Slope:", reg_luas.slope)
print("Intercept:", reg_luas.intercept)
print("R-squared:", reg_luas.rvalue ** 2)
print("p-value:", reg_luas.pvalue)
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    x="sqft_living",
    y="price",
    data=df,
    scatter_kws={"alpha": 0.3},
    line_kws={}
)

plt.title("Regresi Linear Luas Bangunan terhadap Harga Rumah")
plt.xlabel("Luas Bangunan (sqft)")
plt.ylabel("Harga Jual Rumah (USD)")

plt.show()
```

```python
# Regresi linear jumlah kamar terhadap harga
reg_kamar = linregress(df["bedrooms"], df["price"])

print("REGRESI LINEAR: JUMLAH KAMAR DAN HARGA")
print("Slope:", reg_kamar.slope)
print("Intercept:", reg_kamar.intercept)
print("R-squared:", reg_kamar.rvalue ** 2)
print("p-value:", reg_kamar.pvalue)
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    x="bedrooms",
    y="price",
    data=df,
    scatter_kws={"alpha": 0.3},
    line_kws={}
)

plt.title("Regresi Linear Jumlah Kamar terhadap Harga Rumah")
plt.xlabel("Jumlah Kamar Tidur")
plt.ylabel("Harga Jual Rumah (USD)")

plt.show()
```

```python
# Analisis Outlier Menggunakan Metode IQR
variabel_utama = ["price", "sqft_living", "bedrooms"]
hasil_outlier = []
for kolom in variabel_utama:
    Q1 = df[kolom].quantile(0.25)
    Q3 = df[kolom].quantile(0.75)
    IQR = Q3 - Q1
    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR
    jumlah_outlier = (
        (df[kolom] < batas_bawah) |
        (df[kolom] > batas_atas)
    ).sum()
    hasil_outlier.append({
        "Variabel": kolom,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Batas Bawah": batas_bawah,
        "Batas Atas": batas_atas,
        "Jumlah Outlier": jumlah_outlier,
        "Persentase Outlier (%)": (jumlah_outlier / len(df)) * 100
    })

hasil_outlier = pd.DataFrame(hasil_outlier)
hasil_outlier
```
